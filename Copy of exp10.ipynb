```python

import numpy as np
from tensorflow.keras.datasets import fashion_mnist
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 1. Load dataset
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

# 2. Normalize and flatten images
X_train = X_train.reshape(len(X_train), -1) / 255.0
X_test = X_test.reshape(len(X_test), -1) / 255.0

# 3. Reduce training time
X_train, y_train = X_train[:10000], y_train[:10000]
X_test, y_test = X_test[:2000], y_test[:2000]

# 4. Train and predict using three SVM kernels
linear = SVC(kernel='linear')
linear.fit(X_train, y_train)
linear_pred = linear.predict(X_test)

poly = SVC(kernel='poly', degree=3)
poly.fit(X_train, y_train)
poly_pred = poly.predict(X_test)

rbf = SVC(kernel='rbf')
rbf.fit(X_train, y_train)
rbf_pred = rbf.predict(X_test)

# 5. Evaluate each model
def evaluate(y_true, y_pred, name):
    print("\n", name)
    print("-" * 40)
    print("Accuracy :", accuracy_score(y_true, y_pred))
    print("Precision:", precision_score(y_true, y_pred, average='weighted'))
    print("Recall   :", recall_score(y_true, y_pred, average='weighted'))
    print("F1 Score :", f1_score(y_true, y_pred, average='weighted'))

# 6. Display results
evaluate(y_test, linear_pred, "Linear SVM")
evaluate(y_test, poly_pred, "Polynomial SVM")
evaluate(y_test, rbf_pred, "RBF SVM")
```
